```python
import pandas as pd 
import numpy as np

import matplotlib.pyplot as plt 
import seaborn as sns

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression

from sklearn.metrics import r2_score

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report

from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import StandardScaler

from sklearn.preprocessing import OneHotEncoder
oe = OneHotEncoder()
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv(r"C:\Users\mackm\Downloads\heart_disease.csv")
df
```

```python
df.drop_duplicates(inplace = True)
```

```python
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())
```

```python
df.dtypes
```

```python
df.head()
```

```python
if "Age" in df.columns:
    df = df[(df["Age"] >= 0) & (df["Age"] <= 120)]
```

```python
df['Age'].fillna(df['Age'].mode()[0], inplace = True)
df['Gender'].fillna(df['Gender'].mode()[0], inplace = True)
df['Blood Pressure'].fillna(df['Blood Pressure'].mode()[0], inplace = True)
df['Cholesterol Level'].fillna(df['Cholesterol Level'].mode()[0], inplace = True)
df['Exercise Habits'].fillna(df['Exercise Habits'].mode()[0], inplace = True)
df['Smoking'].fillna(df['Smoking'].mode()[0], inplace = True)
df['Family Heart Disease'].fillna(df['Family Heart Disease'].mode()[0], inplace = True)
df['Diabetes'].fillna(df['Diabetes'].mode()[0], inplace = True)
df['BMI'].fillna(df['BMI'].mode()[0], inplace = True)
df['High Blood Pressure'].fillna(df['High Blood Pressure'].mode()[0], inplace = True)
df['Low HDL Cholesterol'].fillna(df['Low HDL Cholesterol'].mode()[0], inplace = True)
df['High LDL Cholesterol'].fillna(df['High LDL Cholesterol'].mode()[0], inplace = True)
df['Alcohol Consumption'].fillna(df['Alcohol Consumption'].mode()[0], inplace = True)
df['Stress Level'].fillna(df['Stress Level'].mode()[0], inplace = True)
df['Sleep Hours'].fillna(df['Sleep Hours'].mode()[0], inplace = True)
df['Sugar Consumption'].fillna(df['Sugar Consumption'].mode()[0], inplace = True)
df['Triglyceride Level'].fillna(df['Triglyceride Level'].mode()[0], inplace = True)
df['Fasting Blood Sugar'].fillna(df['Fasting Blood Sugar'].mode()[0], inplace = True)
df['CRP Level'].fillna(df['CRP Level'].mode()[0], inplace = True)
df['Homocysteine Level'].fillna(df['Homocysteine Level'].mode()[0], inplace = True)
```

```python
numerical_cols = ['Age', 'Blood Pressure', 'Cholesterol Level', 'BMI', 'Sleep Hours',
                  'Triglyceride Level', 'Fasting Blood Sugar', 'CRP Level', 'Homocysteine Level']

for col in numerical_cols:
    plt.figure(figsize=(6, 4))
    sns.histplot(data=df, x=col, kde=True, bins=30, color='skyblue')
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()
```

```python
categorical_cols = ['Gender', 'Smoking', 'Diabetes', 'Family Heart Disease',
                    'High Blood Pressure', 'High LDL Cholesterol',
                    'Alcohol Consumption', 'Stress Level', 'Sugar Consumption']

for col in categorical_cols:
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x=col, hue='Gender', palette='Set2')
    plt.title(f'Heart Disease Status by {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
sns.lineplot(x='Age',y = 'BMI',hue = 'Gender',data = df, style ='Alcohol Consumption',marker = 'o')
plt.xlabel('Age')
plt.ylabel("Alcohol Consumption")
plt.tight_layout()
plt.grid(True)
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
for i in ('Gender','Exercise Habits','Smoking','Family Heart Disease','Diabetes','High Blood Pressure','High LDL Cholesterol','Heart Disease Status','Alcohol Consumption','Stress Level','Sugar Consumption'):
    df[i] = le.fit_transform(df[i])
```

```python
for col in df.select_dtypes(include='object'):
    if set(df[col].dropna().unique()) <= {'Yes', 'No'}:
        df[col] = df[col].map({'Yes': 1, 'No': 0})
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
x = df.drop('Heart Disease Status',axis=1)
y = df['Heart Disease Status']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.3,random_state = 42)
```

```python
lr = LogisticRegression()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
y_pred
```

```python
conf_metrics = confusion_matrix(y_test,y_pred)
print('Confusion matric is: \n',conf_metrics)
```

```python
acc_score = accuracy_score(y_test,y_pred)
print('Accuracy Score is :',acc_score*100)   
```

```python
clss_rep = classification_report(y_test,y_pred)
print('Summary of the model performance : \n',clss_rep)
```
